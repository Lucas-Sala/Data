# Integração das Tabelas

```python
import pandas as pd
```

```python
vendas22 = pd.read_excel("../processed/vendas22.xlsx")
vendas23 = pd.read_excel("../processed/vendas23.xlsx")
vendas24 = pd.read_excel("../processed/vendas24.xlsx")
```

```python
df_vendas = pd.concat(
    [vendas22,vendas23,vendas24],
    ignore_index=True
)
```

```python
df_vendas.info()
```

```python
df_vendas.head()
```

```python
clientes = pd.read_excel("../processed/clientes.xlsx")
lojas = pd.read_excel("../processed/lojas.xlsx")
produtos = pd.read_excel("../processed/produtos.xlsx")
```

```python
df_vendas = df_vendas.merge(
    clientes,
    how="left",
    on="Id Cliente"
)
```

```python
df_vendas.head()
```

```python
df_vendas.info()
```

```python
df_vendas = df_vendas.merge(
    lojas,
    how="left",
    on="Id Loja"
)
```

```python
df_vendas.info()
```

```python
df_vendas.head()
```

```python
df_vendas = df_vendas.merge(
    produtos.drop(columns="Preço Unitário"),
    how="left",
    on="Id Produto"
)
```

```python
df_vendas.info()
```

```python
df_vendas["Marca"] = df_vendas["Marca"].str.strip()
```

```python
df_vendas.to_parquet("../processed/df_Vendas.parquet", index=False)
```
