# Tratamento das Planilhas de Cadastro

```python
import pandas as pd
from src.limpeza import ajusta_nome, substitui_genero, limpar_preco
```

```python
clientes = pd.read_excel("../raw/Cadastro Clientes.xlsx")
lojas = pd.read_excel("../raw/Cadastro Lojas.xlsx")
produtos = pd.read_excel("../raw/Cadastro Produto.xlsx")
```

## Clientes

```python
clientes.info()
```

```python
clientes.shape
```

```python
clientes.isnull().sum()
```

```python
clientes.head()
```

```python
clientes = ajusta_nome(clientes, "Nome Completo")
```

```python
clientes["Nome Cliente"] = clientes["Nome Completo"].str.split(" ", expand=True).apply(lambda x: f"{x[0]}", axis=1)
```

```python
clientes.head()
```

```python
clientes = substitui_genero(clientes, "Genero")
clientes.rename(columns={"Genero":"Gênero"}, inplace=True)
```

```python
clientes.rename(columns={"Data de Nacimento":"Data de Nascimento"}, inplace=True)
```

```python
clientes.head()
```

```python
clientes.isnull().sum()
```

```python
clientes.duplicated().sum()
```

```python
clientes.to_excel("../processed/clientes.xlsx", index=False)
```
